## Objetivo

Entrenar un Perceptrón para distinguir dos especies del conjunto de datos Iris usando únicamente longitud y ancho del pétalo. El foco del experimento es visualizar la frontera lineal aprendida por el modelo.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay
```

```python
iris = load_iris(as_frame=True)
df = iris.frame.copy()

# Problema binario: setosa vs versicolor
df = df[df["target"].isin([0, 1])].copy()

features = ["petal length (cm)", "petal width (cm)"]
X = df[features]
y = df["target"]

df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = Pipeline([
    ("scaler", StandardScaler()),
    ("perceptron", Perceptron(max_iter=2000, eta0=0.1, random_state=42))
])

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print(f"Accuracy de prueba: {accuracy_score(y_test, y_pred):.3f}")
```

```python
def plot_decision_regions(model, X, y, title):
    x1_min, x1_max = X.iloc[:, 0].min() - 0.5, X.iloc[:, 0].max() + 0.5
    x2_min, x2_max = X.iloc[:, 1].min() - 0.3, X.iloc[:, 1].max() + 0.3

    xx1, xx2 = np.meshgrid(
        np.linspace(x1_min, x1_max, 300),
        np.linspace(x2_min, x2_max, 300)
    )
    grid = pd.DataFrame(np.c_[xx1.ravel(), xx2.ravel()], columns=X.columns)
    zz = model.predict(grid).reshape(xx1.shape)

    plt.figure(figsize=(9, 6))
    plt.contourf(xx1, xx2, zz, alpha=0.25)
    for label in sorted(y.unique()):
        mask = y == label
        plt.scatter(
            X.loc[mask, X.columns[0]],
            X.loc[mask, X.columns[1]],
            label=iris.target_names[label],
            edgecolor="black"
        )

    plt.xlabel(X.columns[0])
    plt.ylabel(X.columns[1])
    plt.title(title)
    plt.legend()
    plt.show()

plot_decision_regions(model, X, y, "Frontera de decisión aprendida por el Perceptrón")
```

```python
ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=iris.target_names[:2]
)
plt.title("Matriz de confusión")
plt.show()
```

## Conclusiones

Este experimento muestra cuándo un modelo lineal es suficiente. En estas dos especies y con estas variables, el Perceptrón puede construir una frontera sencilla e interpretable. La visualización también ayuda a entender por qué los problemas no lineales requieren arquitecturas más flexibles.
